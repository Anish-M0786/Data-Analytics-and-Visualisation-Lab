```python
import pandas as pd

# Load the dataset
df = pd.read_csv(
    "data.csv",
    engine="python",
    on_bad_lines="skip"
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Display first and last few rows

print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

# Check data types and general information
print("\nDataFrame Information:")
df.info()

# Summary statistics
print("\nSummary Statistics:")
print(df.describe())
```

```python
# Handle missing values

print("Missing values before handling:")
print(df.isnull().sum())

# Fill missing Rating values with the mean
df["Rating"] = df["Rating"].fillna(df["Rating"].mean())

# Fill missing text values with the mode
df["Current Ver"] = df["Current Ver"].fillna(df["Current Ver"].mode()[0])
df["Android Ver"] = df["Android Ver"].fillna(df["Android Ver"].mode()[0])

print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python
# Create a new column based on an existing column

df["Double_Rating"] = df["Rating"] * 2

print("New column created:")
print(df[["App", "Rating", "Double_Rating"]].head())

# Create a Series from the Rating column

series = df["Rating"]

print("\nSeries:")
print(series.head())

# Basic Series operation
print("\nSeries addition:")
print(series.head() + 10)
```

```python
# Filter rows based on multiple conditions

filtered_df = df[
    (df["Rating"] > 4.5) &
    (df["Reviews"] > 1000)
]

print("Filtered DataFrame:")
print(filtered_df[["App", "Category", "Rating", "Reviews"]].head(10))
```

```python
# Group data by Category and calculate average Rating

grouped = df.groupby("Category")["Rating"].mean()

print("Average Rating by Category:")
print(grouped)
```

```python
# Sort DataFrame by Rating in descending order

df_sorted = df.sort_values(by="Rating", ascending=False)

print("Sorted DataFrame:")
print(df_sorted[["App", "Category", "Rating", "Reviews"]].head(10))
```

```python
# Boolean masking using the median Rating

median_rating = df["Rating"].median()

masked_df = df[df["Rating"] > median_rating]

print("Median Rating:", median_rating)
print("\nMasked DataFrame:")
print(masked_df[["App", "Category", "Rating"]].head(10))
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Remove rows containing missing values
df.dropna(inplace=True)

print("Duplicates removed and missing values dropped.")
print("Remaining shape:", df.shape)
print("\nRemaining missing values:")
print(df.isnull().sum())
```

```python
# Create a new DataFrame with selected columns

subset_df = df[["App", "Category", "Rating", "Reviews"]]

print("Subset DataFrame:")
print(subset_df.head())

# Save the subset DataFrame to a CSV file

subset_df.to_csv("filtered_data.csv", index=False)

print("\nFiltered data saved successfully as filtered_data.csv")
```

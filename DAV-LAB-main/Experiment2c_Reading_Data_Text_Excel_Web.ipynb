```python
import pandas as pd

# Read data from CSV/text file
text_df = pd.read_csv(
    "data.csv",
    engine="python",
    on_bad_lines="skip"
)

print("Text/CSV Data:")
print(text_df.head())
```

```python
# Create sample Excel data

excel_data = {
    "Product": ["Laptop", "Smartphone", "Tablet", "Headphones"],
    "Price": [1000, 800, 500, 100],
    "Quantity": [5, 8, 10, 15]
}

excel_sample = pd.DataFrame(excel_data)

# Save the data as an Excel file
excel_sample.to_excel("data_2c.xlsx", index=False)

print("Excel file created successfully!")
print(excel_sample)
```

```python
# Read data from Excel file

excel_df = pd.read_excel(
    "data_2c.xlsx",
    sheet_name="Sheet1"
)

print("Excel Data:")
print(excel_df.head())
```

```python
# Read data from a web-based CSV source

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"

web_df = pd.read_csv(url)

print("Web Data:")
print(web_df.head())
```

```python
# Check missing values

print("Missing values in Text/CSV data:")
print(text_df.isnull().sum())

print("\nMissing values in Excel data:")
print(excel_df.isnull().sum())

print("\nMissing values in Web data:")
print(web_df.isnull().sum())

# Handle missing values

text_df = text_df.ffill()
excel_df = excel_df.bfill()
web_df = web_df.dropna()

print("\nMissing values handled successfully!")
```

```python
# Save processed data

text_df.to_csv("processed_text.csv", index=False)

excel_df.to_excel("processed_excel.xlsx", index=False)

print("Processed text data saved as processed_text.csv")
print("Processed Excel data saved as processed_excel.xlsx")
```

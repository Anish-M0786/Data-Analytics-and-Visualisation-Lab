```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

print("Libraries imported successfully!")
```

```python
# Load UCI Diabetes Dataset

uci_diabetes = pd.read_csv("uci_diabetes.csv")

print("UCI Diabetes Dataset loaded successfully!")
print("\nDataset Sample:")
print(uci_diabetes.head())
```

```python
# Normal Curve for Glucose

plt.figure(figsize=(6, 5))

sns.histplot(
    uci_diabetes["Glucose"],
    kde=True,
    stat="density",
    linewidth=0
)

x = np.linspace(
    uci_diabetes["Glucose"].min(),
    uci_diabetes["Glucose"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["Glucose"].mean(),
        uci_diabetes["Glucose"].std()
    ),
    'r'
)

plt.xlabel("Glucose")
plt.ylabel("Density")
plt.title("Normal Curve - Glucose")
plt.show()
```

```python
# Normal Curve for BMI

plt.figure(figsize=(6, 5))

sns.histplot(
    uci_diabetes["BMI"],
    kde=True,
    stat="density",
    linewidth=0
)

x = np.linspace(
    uci_diabetes["BMI"].min(),
    uci_diabetes["BMI"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["BMI"].mean(),
        uci_diabetes["BMI"].std()
    ),
    'r'
)

plt.xlabel("BMI")
plt.ylabel("Density")
plt.title("Normal Curve - BMI")
plt.show()
```

```python
print("RESULT:")
print(
    "The normal curves show the distribution of Glucose and BMI, "
    "indicating data spread and skewness."
)
```

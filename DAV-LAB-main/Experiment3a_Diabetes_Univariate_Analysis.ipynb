```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

print("Libraries imported successfully!")
```

```python
# Load Pima Indians Diabetes dataset

pima_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

pima_diabetes = pd.read_csv(
    "pima_diabetes.csv",
    header=None,
    names=pima_columns
)

print("Pima Diabetes Dataset loaded successfully!")
print("Shape:", pima_diabetes.shape)
print("\nPima Indians Diabetes Dataset Sample:")
print(pima_diabetes.head())
```

```python
# Create UCI Diabetes dataset

uci_diabetes = pima_diabetes.sample(
    n=100,
    random_state=42
).reset_index(drop=True)

uci_diabetes.to_csv(
    "uci_diabetes.csv",
    index=False
)

print("UCI Diabetes Dataset created successfully!")
print("Shape:", uci_diabetes.shape)
print("\nUCI Diabetes Dataset Sample:")
print(uci_diabetes.head())
```

```python
# Define relevant numerical columns

numerical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

print("Numerical columns selected:")
print(numerical_columns)
```

```python
# Define univariate analysis function

def univariate_analysis(df, columns):
    stats = {}

    for col in columns:
        stats[col] = {
            "Mean": np.mean(df[col]),
            "Median": np.median(df[col]),
            "Mode": df[col].mode()[0],
            "Variance": np.var(df[col], ddof=1),
            "Standard Deviation": np.std(df[col], ddof=1),
            "Skewness": skew(df[col]),
            "Kurtosis": kurtosis(df[col])
        }

    return pd.DataFrame(stats).T

print("Univariate analysis function created successfully!")
```

```python
# Perform Univariate Analysis

uci_stats = univariate_analysis(
    uci_diabetes,
    numerical_columns
)

pima_stats = univariate_analysis(
    pima_diabetes,
    numerical_columns
)

print("UCI Diabetes Dataset Statistics:")
print(uci_stats)

print("\nPima Indians Diabetes Dataset Statistics:")
print(pima_stats)
```

```python
print("RESULT:")
print(
    "The univariate analysis of the UCI Diabetes and Pima Indians "
    "Diabetes datasets reveals differences in central tendency, "
    "dispersion, and distribution. Variations in skewness and "
    "kurtosis indicate differences in data patterns between the datasets."
)
```

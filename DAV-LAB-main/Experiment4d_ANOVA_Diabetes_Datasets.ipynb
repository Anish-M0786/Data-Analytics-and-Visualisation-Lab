```python
# Import Libraries

import pandas as pd
import numpy as np
from scipy.stats import f_oneway

print("Libraries imported successfully!")
```

```python
# Load the Datasets

uci_diabetes = pd.read_csv("uci_diabetes.csv")

pima_diabetes = pd.read_csv(
    "pima_diabetes.csv",
    header=None,
    names=[
        "Pregnancies",
        "Glucose",
        "BloodPressure",
        "SkinThickness",
        "Insulin",
        "BMI",
        "DiabetesPedigreeFunction",
        "Age",
        "Outcome"
    ]
)

print("UCI Diabetes Dataset loaded successfully!")
print("Pima Indians Diabetes Dataset loaded successfully!")

print("\nUCI Dataset Sample:")
print(uci_diabetes.head())

print("\nPima Dataset Sample:")
print(pima_diabetes.head())
```

```python
# Select Relevant Numerical Columns

numerical_columns = [
    "Glucose",
    "BloodPressure",
    "BMI"
]

print("Selected numerical columns:")
print(numerical_columns)
```

```python
# Perform One-Way ANOVA

anova_results = {}

for col in numerical_columns:
    f_stat, p_value = f_oneway(
        uci_diabetes[col],
        pima_diabetes[col]
    )

    anova_results[col] = {
        "F-statistic": f_stat,
        "P-value": p_value
    }

print("ANOVA calculations completed!")
```

```python
# Convert ANOVA Results to DataFrame

anova_df = pd.DataFrame(anova_results).T

print("ANOVA Results:")
print(anova_df)
```

```python
# Interpret ANOVA Results

alpha = 0.05

for col in numerical_columns:
    p_value = anova_df.loc[col, "P-value"]

    if p_value < alpha:
        print(f"{col}: Significant difference between the datasets.")
    else:
        print(f"{col}: No significant difference between the datasets.")
```

```python
print("\nRESULT:")
print(
    "The One-Way ANOVA compares the means of Glucose, Blood Pressure, "
    "and BMI between the UCI Diabetes and Pima Indians Diabetes datasets. "
    "A p-value less than 0.05 indicates a statistically significant "
    "difference between the dataset means."
)
```
